# XJTU-SY Source Inspection

**Status:** Exploratory  
**Purpose:** 실제 XJTU-SY local source의 존재·inventory·최상위 구조를 확인하고, 첫 Domain Adapter spike에서 답해야 할 질문을 구체화합니다.

이 Notebook은 dataset parser나 canonical contract의 Source of Truth가 아닙니다. 반복 가능한 로직은 `src/industrial_phm/`로 이동하고, 구조 변경 결론은 docs/ADR에 기록합니다.


```python
from __future__ import annotations

import os
from pathlib import Path

from industrial_phm.data.registry import get_dataset
from industrial_phm.data.validation import inspect_source

DATASET_ID = "xjtu-sy"
SOURCE_ENV = "INDUSTRIAL_PHM_XJTU_SOURCE"
source = Path(os.environ.get(SOURCE_ENV, "data/raw/xjtu-sy")).expanduser()
manifest = get_dataset(DATASET_ID)

print(f"dataset: {manifest.dataset_id}")
print(f"title: {manifest.title}")
print(f"provider: {manifest.provider}")
print(f"official source: {manifest.source_url}")
print(f"local candidate: {source}")
```

```python
if source.exists():
    inspection = inspect_source(source)
    print(f"source kind: {inspection.kind}")
    print(f"files: {inspection.file_count}")
    print(f"bytes: {inspection.total_bytes}")
    print("inspection: local inventory only; upstream authenticity is not verified")
else:
    print("XJTU-SY local source was not found.")
    print(f"Set {SOURCE_ENV} or place the dataset at data/raw/xjtu-sy.")
    print("Before continuing, run:")
    print("  uv run industrial-phm data inspect xjtu-sy --source /path/to/XJTU-SY")
```

## Adapter spike에서 확인할 질문

- 현재 `CanonicalTimeSeries` 한 개가 acquisition, run, 또는 다른 granularity 중 무엇과 가장 자연스럽게 대응하는가?


```python
if source.is_dir():
    entries = sorted(source.iterdir(), key=lambda path: path.name)
    print(f"top-level entries: {len(entries)}")
    for path in entries[:50]:
        kind = "dir" if path.is_dir() else "file"
        print(f"{kind:4}  {path.name}")
    if len(entries) > 50:
        print(f"... {len(entries) - 50} more entries")
elif source.exists():
    message = (
        "The configured source is a file; inspect its containing structure "
        "before defining adapter semantics."
    )
    print(message)
```
